```python
from omegaconf import OmegaConf
from data.syn_transform_gnn_inputs import load_synthetic

cfg = OmegaConf.create({
    "data": {
        "name": "syn",
        "target": "DILI",
        "dataset": {
            "root_dir": "/content/drive/MyDrive/synthetic_pharmacotherapy_v2 2",
            "nodes_file": "synthetic_pharmacotherapy_v2_nodes.csv",
            "edges_file": "synthetic_pharmacotherapy_v2_edges.csv",
            "samples": {
                "clean": "synthetic_pharmacotherapy_v2_samples_clean.csv",
                "nooverlap": "synthetic_pharmacotherapy_v2_samples_nooverlap.csv",
                "multihospital": "synthetic_pharmacotherapy_v2_samples_multihospital.csv",
            },
            "scenario": "clean",
        }
    }
})

nodes, edges, samples = load_synthetic(cfg)
```

```python
ENDPOINTS = [
    "AKI", "DILI", "Depression", "Falls", "Delirium",
    "GI_bleeding", "Hyponatremia", "Hyperkalemia",
    "QT_arrhythmia", "Hospitalization",
]

N_total = len(samples)
stats = []

for ep in ENDPOINTS:
    if ep not in samples.columns:
        continue
    N_pos = samples[ep].sum()
    N_neg = N_total - N_pos
    prevalence = N_pos / N_total
    stats.append((ep, int(N_pos), int(N_neg), prevalence))

for ep, N_pos, N_neg, prev in stats:
    print(f"{ep}: pos={N_pos}, neg={N_neg}, prev={prev:.4f}")
```

```python
import pandas as pd

ENDPOINTS = [
    "AKI", "DILI", "Depression", "Falls", "Delirium",
    "GI_bleeding", "Hyponatremia", "Hyperkalemia",
    "QT_arrhythmia", "Hospitalization",
]

# tylko endpointy
E = samples[ENDPOINTS].astype(int)

# macierz współwystępowania: liczba pacjentów z ep_i=1 i ep_j=1
coocc_counts = pd.DataFrame(index=ENDPOINTS, columns=ENDPOINTS, dtype=int)

for ep_i in ENDPOINTS:
    for ep_j in ENDPOINTS:
        coocc_counts.loc[ep_i, ep_j] = ((E[ep_i] == 1) & (E[ep_j] == 1)).sum()

print(coocc_counts)
```
